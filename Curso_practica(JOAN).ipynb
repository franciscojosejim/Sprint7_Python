## Diccionarios:

## Bucles:
- Automatiza un proceso

```python
notas = [4,3,3,7,6,8,7,9,3,2,4,5,6,10]
#por cada nota en las notas que hay, ensena nota:
#for pikachu in notas:
    #print(pikachu)

#for x in notas:
 #   print("La nota es", x)

#Buvles por valor
for nota in notas:
    print("\nLa nota es: ", nota, "\n") # "\n"Salto de linea
    if nota < 5:
        if nota >= 3.5:
             print("Revisaremos el examen")
        else:
            print("Suspenso") #Identado 2 niveles    
    else:
        print("Aprobado")
        if nota >= 9:
            print("Matricula de Honor")
```

```python
# Bucles por posicion (primero posicion 0,1,2,3..) Se usan cuando importa la posicion
# Son mas optimas que los for por valor

for posicion in range(len(notas)):
    print("En la posicion", posicion, "hay un", notas[posicion])

```

## Funciones: creacion
- Trozo de codigo **que hace lo que tu quieras** por ejemplo, una funcion avg

```python
# Creamos fun que recibe un num y lo devuelve la div de la suma de los valores
# con el conteo de valores, es decir, la media.
def media (numeros):
    """
    numeros: lista o similiares
    
    """
    suma = sum(numeros)
    nnumeros = len(numeros)
    return suma/nnumeros 

media([4,5,6,7,1,2,1,2,3,4]) #Invocamos func y enviamos una lista de numeros
```

### Columnas en PANDAS:

```python
del df["year"] #Borra la columna
df["nueva columna"] = "Hola" #Creas una nueva col con "Hola","Hola"...
df["velocidad"] = df["distance"] / df["air_time"] # Creamos columna calculada, ej velocidad.
```

### Seleccionar informacion en PANDAS:

```python
df["air_time"] #Seleccionamos columnas. No es un df, es una serie (col de un df)
df[["air_time", "distance"]] #Seleccionas 2 columnas. Esto si es un df

df[0:5] #Seleccionas por posicion.
```

```python
# Seleccionamos info por posicion
#ILOC[] es raro.

df.iloc[3,2] #Fila 4 y Columna 3
df.iloc[0:5, 0:5] #Primeras 5 filas y primeras columnas
```

```python
# Seleccionamos info por valor/condicion.

df.loc[1:5, ["air_time", "origin"]] #Dame las filas de la 1 a la 5 y las columnas que se llamen tal.
df.loc[df["ait_time"] > 120, ["origin"]] #Dame las filas de vuelos +120min y la colm origen.
df.loc[(df["air_time"] > 120) & (df["arr_delay"] > 30), ["origin"]] #+1 seleccion. "/" para OR
```

```python
# Media del tiempo de vuelo de los aviones de JFK
df.loc[df["origin"] == "JFK", "air_time"].mean() 
```

## Tablas resumen en PANDAS:
- Equivalentes a *tablas dinamicas* en Excel (Pivot Table)
- Df se pueden ordenar con **.sort_values["nom"]**

```python
# Groupby para hacer tablas por filas

# Quiero ver la media de los retrasos agrupados por origen:
df.groupby(["carrier"])[["arr_delay"]].mean() #.sum()/.count()/.describe()/

```

```python
# Crosstab para hacer tablas de doble entrada

pd.crosstab(df["origin"],df["dest"]) #tabla de contingencia. Arriba el destino y a la izq el origen.
```
