Este formulario ha de permitir introducir, por cada candidato, nombre y apellidos, teléfono de contacto, años de experiencia dentro de la empresa y conocimiento de Catalán (0-3), dominio de Excel (0-3) y título académico (0-3).

Planteamiento inicial:

Vamos a crear un programa que nos pida informacion por teclado. Esta informacion son los valores de un candidato. El programa terminara cuando registre 10 candidatos en total.

Los valores de los candidatos deben ser validados para evitar errores. Por ejemplo, que el telefono no tenga menos de 9 caracteres y los rangos de cada valor de catalan, excel y titulo sean del 0 al 3.

De cada candidato, deberemos añadir una formula que multiplica por 1.5 los años de experiencia laboral y suma los numeros de los valores catalan, excel y titulo. Esto tiene como objetivo conseguir un puntuaje total para hacer una clasificacion en base al resultado individual de cada candidato

Queremos que cada vez que anadamos un candidato nuevo nos aparezca sus valores individuales, su resultado de la formula y su clasificacion.

Cuando termine el programa imprimiremos la lista entera de candidatos introducidos y ordenarlos por calsificacion y puntuaje.


Lista de funciones

```python

#Función 1: Obtener datos. Pedimos al usuario inputs y los retornamos.

def obtener_datos():
  print("\nNuevo candidato. Introduce tus datos:")
  nombre = input("Escribe tu nombre completo: ").strip()
  telefono = input("Escribe tu telefono de contacto: ").strip()
  anos_exp = int(input("Cuantos años tienes de experiencia?: "))
  catala = int(input("nivel catalan: 0-Sin conocimiento, 1-B1, 2-B2, 3-C1: "))
  excel = int(input("Dominio de excel: 0-basico, 1-funciones, 2-tablas dinamicas, 3-Basic y macros: "))
  titulo = int(input("Titulo academico: 0-ESO, 1-FP, 2-Uni, 3-Master/Doctorado: "))

  return nombre, telefono, anos_exp, catala, excel, titulo
```

```python
# Función 2: Validar Datos. Procesamos los datos obtenidos en la primera funcion como parametros donde nos devuelva un booleano y un mensaje string.

def validar_datos(nombre, telefono, anos_exp, catala, excel, titulo):
    if not nombre:
      return False, "Nombre no pueden estar vacio"
    if len(telefono) < 9:
      return False, "El telefono tiene que tener 9 numeros"
    if anos_exp < 0:
      return False, "No cuentan años de expriencia negativos"
    if not (0 <= catala <= 3):
      return False, "Catalan debe estar entre 0 y 3"
    if not (0 <= excel <= 3):
      return False, "Excel debe estar entre 0 y 3"
    if not (0 <= titulo <= 3):
      return False, "Titulo academico debe estar entre 0 y 3"

    return True, "Datos Validados"
```

```python
#Función 3: Calcular resultado: calcular los datos ya validados segunla formula del enunciado.

def calcular_resultado(anos_exp, catala, excel, titulo):

  return round(anos_exp * 1.5 + catala + excel + titulo, 2)
```

```python
  #Función 4: Clasificar Candidato:
  #Aqui sacamos la categoria segun puntiacion (calcular_resultado())

def clasificar_candidato(puntuacion):
   if puntuacion >= 12:
      return "Prioridad Alta"
   elif 8 <= puntuacion < 12:
      return "Prioridad Media"
   else: return "No prioritario"
```

```python
# Función 5: muestra individual. Imprime resultados del nuevo candidato a añadir

def muestra_indiv(nombre: str, telefono: str, puntuacion: float, categoria: str):

  print("candidato registrado:")
  print(f"\nNombre: {nombre}")
  print(f"telefono: {telefono}")
  print(f"puntuacion: {puntuacion}")
  print(f"categoria: {categoria}")
```

```python
# Función 6: Mostrar informes ordenados por categoria y puntuacion:

def mostrar_informes(candidatos : List[Dict]):
  candidatos_ordenats = sorted(candidatos, key=lambda persona: (persona["categoria"], persona["puntuacion"]), reverse=True)

  print("\nInforme Final de 10 Candidatos:")
  for posicion, persona in enumerate(candidatos_ordenats, 1):
        print(f"{posicion}. {persona['nombre']} - telefono: {persona['telefono']} - {persona['puntuacion']} puntos ({persona['categoria']})")

```

```python
#Función principal main

def main():

  candidatos = []
  print("Bienvenido a la seleccion de personal!")

  while len(candidatos) < 10:
    print(f"\nCandidato {len(candidatos)+1} de 10:")


    #obtengo datos
    nombre, telefono, anos_exp, catala, excel, titulo = obtener_datos()
    #valido datos
    valido, mensaje = validar_datos(nombre, telefono, anos_exp, catala, excel, titulo)

    if valido:

      puntuacion = calcular_resultado(anos_exp, catala, excel, titulo)
      categoria = clasificar_candidato(puntuacion)
      categoria_num = 3 if categoria == "Prioridad Alta" else 2 if categoria == "Candidato Adecuado" else 1

      candidatos.append({
      "nombre" : nombre,
      "telefono" : telefono,
      "puntuacion": puntuacion,
      "categoria": categoria,
      "categoria_num" : categoria_num
      })


      muestra_indiv(nombre, telefono, puntuacion, categoria)
    else:
      print(f"Error: {mensaje}. Intentalo1 de nuevo.")

  mostrar_informes(candidatos)


main()





```

```python
#1 Función para obtener el factor de la edad segun rango de edades (clientes(grup_edat)). Usamos GET para evitar errores??
def obtener_factor_edad(grup_edat):
  return factors_edat[grup_edat]
```

```python
#2 Función para segmentar (+mensaje) y factor de comportamiento.

def obtener_segmento(codi_comportament):
  factor = factors_comportament[codi_comportament]
  segmento = segments[codi_comportament]
  return segmento["segment"], segmento["missatge"], factor
```

```python
#3 Función para determinar el riesgo. Recibimos dos parametros y devolvemos un resultado redondeado.

def calcular_riesgo(factor_comp, factors_edat):
  return round(factor_comp * factors_edat, 2)
```

```python
#4 Función para añadir toda la info calculada previamente a una nueva lista de clientes. Le pasamos una lista y nos la devuelve enriquecida.
def procesar_clientes(clients):
  clientes_procesados = []

  for cliente in clients:
    factor_edat = obtener_factor_edad(cliente["grup_edat"])
    segment, missatge, factor_comp = obtener_segmento(cliente["codi_comportament"])
    riesgo_total = calcular_riesgo(factor_comp, factor_edat)

    cliente_copia = cliente.copy()
    cliente_copia.update({
      "factor_edat": factor_edat,
      "segment": segment,
      "missatge": missatge,
      "factor_comportament": factor_comp,
      "riesgo_total": riesgo_total
    })

    clientes_procesados.append(cliente_copia)

  return clientes_procesados
```

```python
#5 Función para mostrar df
def crear_dataframe(clientes_procesados):
  df = pd.DataFrame(clientes_procesados)
  return df
```

```python

#6 Función para responder a las preguntas
def mostrar_respuestas(df):
  print("\n1) ¿Qué clientes se deben contactar primero en una campaña de retargeting?")
  top_riesgo = df.nlargest(3, "riesgo_total")[["nom","segment", "riesgo_total"]]
  print(top_riesgo.to_string(index=False))

  print("\n2) ¿Qué diferencias hay entre clientes con el mismo comportamiento pero de edades diferentes? (Por ejemplo: ¿un cliente joven que abandona el carrito de compra hace 24 h tiene el mismo riesgo que uno de mayor edad?")
  print("Ejemplo joven (con factor 4)", df[df["nom"]=="Laura Serra"]["riesgo_total"].values[0])
  print("Ejemplo no tan joven (con factor 4)", df[df["nom"]=="Mónica Pérez"]["riesgo_total"].values[0])
  print("SOLUCION: Los jovenes tienen mas riesgo por que su factor de edad es mas alto")

  print("\n3) ¿Qué clientes muestran tan poco interés que no vale la pena invertirles presupuesto?")
  #los clientes que tienen 0 en codi comportament
  poco_interes = df[df["codi_comportament"] == 0].nsmallest(5, "riesgo_total")[["nom", "segment", "missatge", "riesgo_total"]]
  print(poco_interes.to_string(index=False))

  print("\n4) ¿Qué cliente sería el más rentable de recuperar?")
  #aquellos que tengan menos riesgo
  millor = df.loc[df["riesgo_total"].idxmax()]
  print(f"El mejor cliente es {millor['nom']} con un riesgo de {millor['riesgo_total']}")

```

Planteamiento inicial:

Tenemos una lista de pacientes y un diccionario con valores de riesgo segun sintoma.
La lista de pacientes contiene informacion mal estructurada donde se puede localizar nombre, edad seguido de una serie de sintomas, tambien mal estructurados (con mayusculas, sin espacios...)

Con estos datos, se nos pide calcular el riesgo sumando los valores de niveles de riesgo por cada uno de los sintomas de la lisa pacientes.

Por ultimo ordenaremos la nueva lista con el total de puntos por sintoma agrupado por paciente de mayor a menor riesgo con el objetivo de priorizar la atencion de la forma mas efectiva.

Lista de funciones:

2) Función para normalizar el texto

```python
pacientes = [
    "Maria|45|tos, Fiebre Alta, dolor_pecho",
    "Luis|33|dolor_cabeza,faTiga",
    "Sara|67|fiebre alta, dificultad RESPIRAR, tos",
    "Jordi|52|fatiga , tos",
    "Anna|29|fiebre alta,dolor_cabeza"
]

niveles_riesgo = {
    "fiebre_alta": 3,
    "dificultad_respirar": 5,
    "dolor_pecho": 4,
    "tos": 1,
    "fatiga": 1,
    "dolor_cabeza": 1
}

```

```python
#1) Convertir cada string en una estructura limpia (diccionario o similar)
# para ello, convetimos string de paciente en un diccionario con .split() y cada posicion le damos una clave
def procesar_pacientes(registro: [str]) -> Dict:
  partes = registro.split("|")
  nombre = partes[0].strip()
  edad = int(partes[1].strip())
  sintomas_raw = partes[2].strip().split(",")
  sintomas_limpios = [procesar_sintomas(sintoma) for sintoma in sintomas_raw]

  return {"nombre": nombre,
          "edad": edad,
          "sintomas": sintomas_limpios
          }

```

```python
# 2)Separar y normalizar la lista de síntomas; min, espacios en blanco y dividir por comas. Reemplazamos espacios en blanco por "_" para que coincida con "niveles_riesgo"

def procesar_sintomas(sintoma: str) -> str:

  return sintoma.strip().lower().replace(" ", "_").replace("-", "_")

```

```python

# 3) Asignar un numero riesgo a cada síntoma utilizando el diccionario “niveles_riesgo” y devolver el total en integrer.

def asignar_riesgo(sintomas: List[str], niveles_riesgo: Dict) -> int:
  riesgo = 0
  for sintoma in sintomas:
    riesgo += niveles_riesgo.get(sintoma, 0)

  return riesgo
```

```python
#4)Ordena los pacientes de mayor a menor riesgo para determinar la prioridad de atención.
# Enviamos dos parametros (esperamos una lista y un diccionario), lo recorremos, los anadimos a una lista vacia y devolvemos la lista ordenada con el tipo dataframe.

def ordenar_pacientes(pacientes_riesgo: List[str], niveles_riesgo : Dict):
  datos = []

  for pacientes in pacientes_riesgo:
    paciente = procesar_pacientes(pacientes)
    riesgo_total = asignar_riesgo(paciente["sintomas"], niveles_riesgo)

    datos.append({"nombre": paciente["nombre"],
                  "edad": paciente["edad"],
                  "sintomas": paciente["sintomas"],
                  "riesgo_total": riesgo_total
                  })

  df = pd.DataFrame(datos)
  df_ordenado = df.sort_values(by="riesgo_total",
                                 ascending=False).reset_index(drop=True)

  return df, df_ordenado
```

```python
#5 Función principal
def main():
  df_pacientes, df_prioridad = ordenar_pacientes(pacientes, niveles_riesgo)

  #Opcion de imprimir la lista sin ordenar.
  #print("Datos procesados")
  #print(df_pacientes)
  print("Datos ordenados, mayor riesgo primero!")
  print(df_prioridad[["nombre", "edad", "sintomas", "riesgo_total"]])

```

```python
clients = [
    {"nom": "Ana", "edat": 42, "codi": 0},
    {"nom": "Carlos", "edat": 30, "codi": 2},
    {"nom": "Isabel", "edat": 55, "codi": 3},
    {"nom": "Jorge", "edat": 40, "codi": 1},
    {"nom": "Marta", "edat": 28, "codi": 0}
]

nivells_seguro = {
    0: {"categoria": "Baix risc", "preu_base": 120, "fraude": 0.01},
    1: {"categoria": "Risc mitjà", "preu_base": 200, "fraude": 0.05},
    2: {"categoria": "Risc alt", "preu_base": 350, "fraude": 0.15},
    3: {"categoria": "Risc crític", "preu_base": 500, "fraude": 0.30}
}
```

```python
#1)Clasifica cada cliente utilizando niveles de seguro. Recibo dos parametros y devuelvo un diccionario

def clasificar_cliente(clients, nivells_seguro):
   codi = clients["codi"]
   info = nivells_seguro[codi]

   preu_final = info["preu_base"] * (1 + clients["edat"]/100)

   return {
    "nom": clients["nom"],
    "edat": clients["edat"],
    "categoria": info["categoria"],
    "preu_base": info["preu_base"],
    "probabilidad_fraude": info["fraude"],
    "preu_final": round(preu_final,2)
}
```

```python
#2 Añade la categoría, el precio base y la probabilidad de fraude. Devuelve un dataframe

def procesar_clientes(clients, nivells_seguro):

  datos_procesados = []

  for persona in clients:
    client_info = clasificar_cliente(persona, nivells_seguro)
    datos_procesados.append(client_info)
  df = pd.DataFrame(datos_procesados)
  return df
```

```python
#3 Muestra la tabla completa
def mostrar_tabla_completa(df):
  print("Tabla de Clientes completa")
  print(df)
```

```python
#4 Muestra el dataframe ordenado por  fraude
def mostrar_porfraude(df):
  df_frau = df.sort_values(by="probabilidad_fraude", ascending=False)
  print("\nTabla ordenada por Probabilidad de fraude (des)")
  print(df_frau[["nom", "edat", "categoria", "preu_base", "probabilidad_fraude", "preu_final"]])
```

```python
#5 Muesta el dataframe ordenado por precio
def mostrar_porprecio(df):
  df_preufinal = df.sort_values(by="preu_final", ascending=False)
  print("\nTabla ordenada por Precio final")
  print(df_preufinal[["nom", "edat", "categoria", "probabilidad_fraude", "preu_final"]])
```

```python
#6 Muestra el dataframe con la linea de la persona con mas posibilidad de fraude y precio mas alto que convendria revisar manualmente.
def mostrar_prioridades(df):

  client_mes_fraudulent = df.loc[df["probabilidad_fraude"].idxmax()]
  client_preumesalt = df.loc[df["preu_final"].idxmax()]

  print(f"\nClient amb mes probabilitat de frau: {client_mes_fraudulent["nom"]}"
        f"({client_mes_fraudulent["probabilidad_fraude"]*100:.1f}% -- {client_mes_fraudulent["categoria"]})")

  print(f"\nClient amb preu final mes alt: {client_preumesalt["nom"]}"
        f"({client_preumesalt["preu_final"]}euros -- {client_preumesalt["categoria"]})")
```

```python
def main():
  df_clients = procesar_clientes(clients, nivells_seguro)
  mostrar_tabla_completa(df_clients)
  mostrar_porfraude(df_clients)
  mostrar_porprecio(df_clients)
  mostrar_prioridades(df_clients)
```

```python
main()
```
